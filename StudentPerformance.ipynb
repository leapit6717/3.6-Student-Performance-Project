# Student Performance Analysis
Student ID: leapit6717  
Date: October 24, 2026

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("StudentsPerformance.xlsx")

print(df.head())
print("\nMissing values:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(9, 5))

plt.hist(
    df["math score"],
    bins=range(0, 101, 10),
    color="skyblue",
    edgecolor="black"
)

plt.title("Math Score Distribution - leapit6717")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")
plt.xticks(range(0, 101, 10))
plt.tight_layout()
plt.show()
```

```python
average_math = df.groupby("race/ethnicity")["math score"].mean()

ax = average_math.plot(
    kind="bar",
    figsize=(9, 5),
    color="lightcoral",
    edgecolor="black"
)

for container in ax.containers:
    ax.bar_label(container, fmt="%.2f", padding=3)

plt.title("Average Math Score by Race/Ethnicity - leapit6717")
plt.xlabel("Race/Ethnicity Group")
plt.ylabel("Average Math Score")
plt.xticks(rotation=0)
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.scatter(
    df["math score"],
    df["writing score"],
    color="purple",
    alpha=0.5
)

plt.title("Math and Writing Scores - leapit6717")
plt.xlabel("Math Score")
plt.ylabel("Writing Score")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.scatter(
    df["math score"],
    df["reading score"],
    color="teal",
    alpha=0.5
)

plt.title("Math and Reading Scores - leapit6717")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.tight_layout()
plt.show()
```

```python
scores = df[["math score", "reading score", "writing score"]]
correlations = scores.corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlations,
    annot=True,
    fmt=".3f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Score Correlations - leapit6717")
plt.xlabel("Score Type")
plt.ylabel("Score Type")
plt.tight_layout()
plt.show()
```
